```python
import pandas as pd
```

```python
filename = "log.2026-09-09"
```

```python
with open("log.2026-09-09", "r") as f:
    log_data = f.read()
log_lines = log_data.splitlines()

# extract the fields, parse the key=value pairs and convert to a DataFrame, and add the timestamps (first 2 columns)
log_dicts = []
for line in log_lines:
    entry = {}
    parts = line.split()
    if len(parts) >= 2:
        entry["timestamp"] = f"{parts[0]} {parts[1]}"  # first 2 columns are date + time
        items = parts[2:]
    else:
        items = []

    for item in items:
        if "=" in item:
            key, value = item.split("=", 1)
            entry[key] = value

    log_dicts.append(entry)

log_df = pd.DataFrame(log_dicts)
log_df["timestamp"] = pd.to_datetime(log_df["timestamp"])
log_df = log_df.set_index("timestamp")

# ensure numerica fields are numeric, drop the "%" from cpu and the "M" sufficx from swap, rss, cma_free and cache 
# and drop the C suffix from temp
for col in log_df.columns:
    if col == "cpu":
        log_df[col] = log_df[col].str.rstrip("%").astype(float)
    elif col in ["swap", "rss", "cma_free", "cache"]:
        log_df[col] = log_df[col].str.rstrip("M").astype(float)
    elif col == "temp":
        log_df[col] = log_df[col].str.rstrip("C").astype(float)
```

```python
log_df['cma_free'].plot(figsize=(12, 6), title='CMA Free Memory Over Time')
```
